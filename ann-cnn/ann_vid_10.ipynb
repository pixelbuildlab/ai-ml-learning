```python
x_test=x_test.reshape(10000, 784)
```

```python
x_train=x_train.astype('float32')
```

```python
x_test=x_test.astype('float32')
```

```python
x_train=x_train/255
x_test=x_test/255
```

```python
help(to_categorical)
```

```python
y_train[0], y_test[0],
```

```python
y_train = to_categorical(y_train, np_classes)
y_test  = to_categorical(y_test, np_classes)
```

```python
# # help(Dense)from tensorflow.keras.layers import Input, Dense
# from tensorflow.keras.layers import Input, Dense
# from tensorflow.keras.models import Sequential
```

```python
model= Sequential()
```

```python
model.add(Input(shape=(784,)))
```

```python
model.add(Dense(512))
```

```python
model.add(Activation(activations.relu))
```

```python
model.add(Dropout(0.2))
```

```python
model.add(Dense(512))
```

```python
model.add(Activation(activations.relu))
```

```python
model.add(Dropout(0.2))
model.add(Dense(10))
model.add(Activation(activations.softmax))
```

```python
model.layers
```

```python
optimizer= RMSprop()
```

```python
from keras import metrics
```

```python
model.compile(optimizer, loss_fun.categorical_crossentropy, metrics=[metrics.CategoricalAccuracy()])
```

```python
model.fit(x_train, y_train, batch_size, np_epoch, "auto", validation_data=(x_test, y_test), )
```
