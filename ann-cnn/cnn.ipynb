```python
import numpy as np
```

```python
np.random.seed(1337)
```

```python
from keras.datasets import mnist
from keras.models import Model, Sequential
from keras import activations
from keras.layers import Dropout, Conv2D, MaxPool2D, Activation, Input, Flatten, Dense
from keras.metrics import CategoricalAccuracy
from keras.losses import categorical_crossentropy
from keras.optimizers import RMSprop, AdamW, SGD
```

```python
from keras.utils import to_categorical
```

```python
batch_size=128
nb_classes=10
epoch = 5
```

```python
# Convolution Config
# size, step
img_rows, img_col = 28, 28
nb_filters = 32
# nb_kernel = 3
kernel_size = (3,3)
# MaxPool Config
# size -> (x,y)
nb_pool = (2,2)
# Dropout Config
# percent
# 0.25, .50
```

```python
dataset  = mnist.load_data()
len(dataset)
```

```python
data_train, data_test = dataset

x_test, y_test   =  data_test
x_train, y_train =  data_train
```

```python
x_train.shape[0], "---", y_train.shape
```

```python
x_test.shape[0], "---", y_test.shape
```

```python
x_train = x_train.reshape(x_train.shape[0], img_rows, img_col,1)
x_test = x_test.reshape(x_test.shape[0], img_rows, img_col, 1)
```

```python
x_test = x_test.astype("float32")
x_train = x_train.astype("float32")

x_test = x_test / 255
x_train = x_train / 255
```

```python
y_test = to_categorical(y_test, nb_classes)
y_train = to_categorical(y_train, nb_classes)
```

```python
np.argmax(y_test[0])
```

```python
model = Sequential()
```

```python
model.add(Input(shape=(img_rows, img_col, 1)))
```

```python
model.add(Conv2D(nb_filters,kernel_size=kernel_size, padding='valid'))
model.add(Activation(activation=activations.relu))

model.add(Conv2D(nb_filters, kernel_size=kernel_size))

model.add(Activation(activation=activations.relu))

model.add(Conv2D(nb_filters, kernel_size=kernel_size))

model.add(Activation(activation=activations.relu))

model.add(MaxPool2D(pool_size=nb_pool))

model.add(Dropout(0.25))

model.add(Flatten())

model.add(Dense(128))

model.add(Activation(activation=activations.relu))

model.add(Dropout(0.5))


# Final Layer
model.add(Dense(nb_classes))
model.add(Activation(activation=activations.softmax))
```

```python
model.summary()
```

```python
opt =  RMSprop()
model.compile(opt, categorical_crossentropy, metrics=[CategoricalAccuracy()])
```

```python
x_test.shape, "---", y_test.shape
```

```python
model.fit(x_train, y_train, epochs=epoch, batch_size=batch_size, validation_data=(x_test, y_test))
```

```python
scores = model.evaluate(x_test, y_test)
```
