```python
from numpy.linalg import norm
```

```python
help(norm)
```

```python
A_norm = norm(A)
B_norm = norm(B)
```

```python
A_norm, B_norm
```

```python
numerator = np.dot(A,B)
```

```python
numerator
```

```python
denim= A_norm*B_norm
```

```python
denim
```

```python
cos_sim = numerator/denim
```

```python
cos_sim
```

```python
cos_dis = 1 - cos_sim
```

```python
cos_dis
```

```python
# cos_dis is between 0-2
# 0 mean low distance, Matrix are identical or in same direction
# 1 means perpendicular points
# 2 means total opposite metric
```

```python
VecA = np.random.normal(10,size=2)
VecB = np.random.normal(10,size=2)
```

```python
VecA,VecB
```

```python
import matplotlib.pyplot as plt
```

```python
# Plot lines from (0,0) to the points
plt.plot([0, VecA[0]], [0, VecA[1]], label="VecA", marker="o")
plt.plot([0, VecB[0]], [0, VecB[1]], label="VecB", marker="o")

# Keep aspect ratio square to see the true angle accurately
plt.gca().set_aspect("equal")
plt.legend()
plt.show()
```

```python
def find_similarity(a,b):
    # num
    # A.B
    # dem
    # |A|.|B|
    # res = num/dem
    num = np.dot(a,b)
    dem = norm(a) * norm(b)
    return num/dem 
```

```python
find_similarity(A,B)
```

```python
def find_dist(sim):
    return 1 - sim
```

```python
sim = find_similarity(VecA,VecB)
sim
```

```python
dist = find_dist(sim)
dist
```

```python
round(dist, 3)
```
