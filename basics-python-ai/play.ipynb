```python
# 1. Install plotting dependency directly in the notebook

import numpy as np
import matplotlib.pyplot as plt

# 2. Define the original 2D XOR data points
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 0])

# 3. Create the plot
plt.figure(figsize=(6, 6))

# Plot Class 0 points (Red squares)
plt.scatter(X[y == 0, 0], X[y == 0, 1], color='red', marker='s', s=200, label='Class 0 (Output: 0)')

# Plot Class 1 points (Blue circles)
plt.scatter(X[y == 1, 0], X[y == 1, 1], color='blue', marker='o', s=200, label='Class 1 (Output: 1)')

# 4. Formatting the chart
plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.xticks([0, 1])
plt.yticks([0, 1])
plt.xlabel('Input X1', fontsize=12)
plt.ylabel('Input X2', fontsize=12)
plt.title('The XOR Problem: Non-Linearly Separable', fontsize=14, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='center', fontsize=10)

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from scipy.signal import convolve2d

# Load image
img = mpimg.imread("cat.png")

# If RGB, convert to grayscale
if img.ndim == 3:
    img = img.mean(axis=2)

kernel = [
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
]

sharpened = convolve2d(img, kernel, mode="same", boundary="symm")

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(sharpened, cmap="gray")
plt.title("Sharpened")
plt.axis("off")

plt.show()
```
